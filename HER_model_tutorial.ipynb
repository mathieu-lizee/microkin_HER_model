### Single step reaction with mass transport limitations
- The diffusion coefficient at 300K is set to 1e-9 by default. Another value can be set in 'D_reactant'. Its temperature-dependency is then set by the temperature-dependent viscosity of water following the Stokes-Einstein equation. In the fully mass transported limited regime (DRC1=0 & DRC2=0), the apparent activation energy is that of water's viscosity (14 kJ/mol)
- The setup can be chosen either as RDE with rotation rate given in RPM or as UME with a radius in m. Planar diffusion can be obtained with the RDE setup at omega_RPM=0
- In the resulting Figure, the surface concentration of the reactant is plotted on panel c as a function of overpotential
- the DRC of the 1st step drops (last panel) with increasing mass transport limitation

```python
params_default={'N_steps':1,'step1':'BV','step2':'Heyrovsky',
                'Ea1': 30e3,'k1_0':1e2,'alpha':0.5,
                'mass_transport':True, 'c0': 0.2,'D_reactant':1e-9,
                'setup':'RDE','omega_RPM':0.01,'R': 1e-7,'T': 300}
(eta_tab,current_tab,theta_tab,Ea_tab,A_tab)=plot_analysis(params_default,eta_max=0.3,N_eta=20,temps=np.linspace(280,340,4),titre='')
```

### 2-step model of the HER
- The 2nd step is described either as a Heyrovsky or as a Tafel step. Its activation energy and preexponential factors are fixed as k_2_0 and Ea2
- The binding enthalpy of the intermediate is E_H_ads in eV.
- The free energy of the final state is zero by default but it can be tuned through E_H2 (in eV)

### Volmer-Heyrovsky

```python
params_default={'N_steps':2,'step1':'BV','step2':'Heyrovsky','E_H_ads':-0.1,'alpha':0.5,
                'T': 300, 'Ea1': 10e3,'Ea2':40e3,'k1_0':1e0, 'k2_0': 1e6,'c0': 0.2,
                'mass_transport':False}
(eta_tab,current_tab,theta_tab,Ea_tab,A_tab)=plot_analysis(params_default,eta_max=0.3,N_eta=20,temps=np.linspace(280,340,4),titre='')
```
